Q1 Stop sending emails after 3 users (Using for loop and break)

users = ["Ali", "Sara", "John", "Zara", "Mike"]

Output:

Email sent to Ali

Email sent to Sara

Email sent to John

Limit reached. Stop sending emails.

```python

users = ["Ali", "Sara", "John", "Zara", "Mike"]

for i in range(len(users)):
    if i == 3:
        print("Limit reached. Stop sending emails.")
        break
    print(f"Email sent to {users[i]}")
```

Q2. Stop monitoring temperature if critical temperature reached (Using for loop and break)

temps = [22, 25, 30, 45, 28] critical = 40

Ouput:

Temperature: 22

Temperature: 25

Temperature: 30

Critical temperature reached! Stop monitoring.

```python

temps = [22, 25, 30, 45, 28]

for i in range(len(temps)):
    if temps[i] >= 40:
        print("Critical temperature reached! Stop monitoring.")
        break
    else:
        print(f"Temperature: {temps[i]}")
```

Q3. Calculate total price of items in a shopping cart (Using For loop)

cart = [100, 250, 50, 300]

Output:

Total Price: 700

```python
cart = [100, 250, 50, 300]
total_price = 0

for i in cart:
    total_price += i

total_price
```

Q4. Skip customers with unpaid bills (Using For loop and continue)

customers = [ {"name": "Ali", "paid": True},

{"name": "Sara", "paid": False},

{"name": "John", "paid": True}
]

Output:

Access granted to: Ali

Access granted to: John

```python
customers = [ {"name": "Ali", "paid": True},

{"name": "Sara", "paid": False},

{"name": "John", "paid": True} ]

for i in customers:
    if not i["paid"]:
        continue
    print(f"Access granted to: {i["name"]}")
```

Q6. Extract domain names from email list (Using for loop)

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

Output:

domains: ['gmail.com', 'yahoo.com', 'hotmail.com']



```python
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
```

```python
domains = []

for i in emails:
    domain = i.split("@")[1]
    domains.append(domain)

domains
```

Q7. Create a list of numbers, generate a new list containing only the numbers that are divisible by 3 or 5. (List Comprehension)

```python
num_list = [10 , 14, 15, 45, 35, 33, 23, 54]
divisible_list = [i for i in num_list if i % 3 == 0 or i % 5 == 0]
divisible_list
```

Q8. Create a multiplication table using a nested for loop.

numbers = [2, 3, 4]
multipliers = [1, 2, 3]

Expected Output:

2 x 1 = 2
2 x 2 = 4
2 x 3 = 6
3 x 1 = 3
3 x 2 = 6
3 x 3 = 9
4 x 1 = 4
4 x 2 = 8
4 x 3 = 12

Use a nested for loop.

```python
numbers = [2, 3, 4] 
multipliers = [1, 2, 3]
for i in numbers:
    for j in multipliers:
        print(f"{i} * {j} = {i*j}")
```

Q9. Keep asking the user for a password until the correct password is entered.

correct_password = "python123"

Expected behaviour:

Enter password: abc
Incorrect password. Try again.

Enter password: 12345
Incorrect password. Try again.

Enter password: python123
Access granted!

Use a while loop.

```python
correct_password = "python123"
is_matched = False

while not is_matched:
    user_pass = input("Enter the password: ")

    if user_pass == correct_password:
        is_matched = True
        print("Access granted!")
    else:
        print("Incorrect password. Try again.")
```
